```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

kashti = sns.load_dataset("titanic")
sns.boxplot(x="survived",
            y="age", showmeans=True,
            meanprops= {"marker":"*",
                        "markersize":"12",
                        "markeredgecolor":"red"},
            data= kashti)
plt.xlabel("How many survived", size=10, weight='bold')
plt.ylabel("Age(years)", size=10, weight='bold')
plt.title("Box plot of kitnay doobay or kitnay bach gaye", size=14, weight='bold')

plt.show()  # facet wrap and facet grid
```

```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt                      #using catplot for "facet wrap"

kashti = sns.load_dataset("titanic")      
sns.catplot(x="survived",
            y="age", data= kashti, col= "sex", kind="box", col_wrap=2)
plt.show()
```

```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt                       #using catplot for "facet wrap"

kashti = sns.load_dataset("titanic")      
sns.facegrid(x="survived",
            y="age", data= kashti, col= "sex", kind="box", col_wrap=2)
plt.show()
```
