```python
## 1 Dimension
```

```python
import numpy as np 
a = np.array([1,2,3,4,5,6])
a
```

```python
b = np.zeros(2)
b
```

```python
c = np.ones(3)
c
```

```python
# Create empty array with 2 elements
d = np.empty(3)
d
```

```python
# with range of elements
e= np.arange(6)
e
```

```python
# with specific range of elements
f = np.arange(2,20)
f
```

```python
# Continue...
g = np.arange(2,20,2)
g
```

```python
#linerly space arrays
h = np.linspace(0,10,num=5) # gives use 5 nums
h
```

```python
 # specific data types in array
i = np.ones(5, dtype=np.int8)
i
```

```python
j = np.ones(3, dtype=np.float64)
j
```

```python
## 2 Dimension
```

```python
np.zeros((3,4))
```

```python
np.ones((5,5))
```

```python
np.empty((3,4))
```

```python
## 3 Dimension
```

```python
# making and reshaping a 3d array
c = np.arange(24).reshape(2,3,4)
c
```
