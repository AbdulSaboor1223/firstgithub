## 3- Dictionaries
- An ordered collection of Elements
- key and value
- Curley braces/Brackets{}
- Mutateable/Change the values

```python
#Food and their prices
food1 = {"Samosa":45, "Pakora":200, "Raita":30, "Salad":50}
food1
```

```python
type(food1)
```

```python
keys1 = food1.keys()
keys1
```

```python
values1 = food1.values()
values1
```

```python
#adding new elements
food1["Tikki"]= 10
food1
```

```python
# update the values
```

```python
food1["Tikki"]=15
food1
```

```python
food2 = {"Dates":200, "Choclates":100,  "Cold Drink":200, "Nimko":200}
food2
```

```python
food1.update(food2)
food1
```
